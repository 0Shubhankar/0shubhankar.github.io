```python
import osmnx as ox
import geopandas as gpd
import pandas as pd

# Dictionary with city names and their approximate central coordinates
cbd_coordinates = {
    "Delhi, India": (28.6315, 77.2167),
    "Mumbai, India": (18.9250, 72.8246),
    "Bengaluru, India": (12.9758, 77.6069),
    "New York, USA": (40.7074, -74.0113),
    "Los Angeles, USA": (34.0522, -118.2437),
    "Chicago, USA": (41.8781, -87.6298),
    "Shanghai, China": (31.2354, 121.4977),
    "Beijing, China": (39.9141, 116.4551),
    "Shenzhen, China": (22.5412, 114.0579),
    "Istanbul, Turkey": (41.0798, 29.0150),
    "Moscow, Russia": (55.7495, 37.5371),
    "London, UK": (51.5155, -0.0922),
    "Saint Petersburg, Russia": (59.9311, 30.3558),
    "Berlin, Germany": (52.5096, 13.3763),
    "Madrid, Spain": (40.4469, -3.6937),
    "Kinshasa, DRC": (-4.3147, 15.3060),
    "Lagos, Nigeria": (6.4527, 3.3903),
    "Cairo, Egypt": (30.0444, 31.2357),
    "Paris, France": (48.8924, 2.2374),
    "Tokyo, Japan": (35.6813, 139.7640),
    "Mexico City, Mexico": (19.4326, -99.1332),
    "São Paulo, Brazil": (-23.5614, -46.6559),
    "Buenos Aires, Argentina": (-34.6037, -58.3816),
    "Rio de Janeiro, Brazil": (-22.9083, -43.1773),
    "Sydney, Australia": (-33.8688, 151.2093),
    "Melbourne, Australia": (-37.8136, 144.9631),
    "Auckland, New Zealand": (-36.8485, 174.7633)
}

# Tags for different types of parking
parking_tags = {'amenity': ['bicycle_parking', 'motorcycle_parking', 'parking', 
                            'parking_entrance', 'parking_space'], 'building': 'parking'}

# Dictionary to store results
parking_results = {}

# Function to calculate area in square meters of parking spaces
def calculate_parking_area(gdf):
    gdf = gdf.to_crs(epsg=3395)  # Convert to metric CRS for area calculation
    return gdf['geometry'].area.sum()

# Loop through each city
for city, (lat, lon) in cbd_coordinates.items():
    try:
        # Get the city's boundary using OSMnx
        city_boundary_gdf = ox.geocoder.geocode_to_gdf(city)
        city_boundary = city_boundary_gdf.unary_union

        # Query OSM for parking within the city boundary
        parking_gdf = ox.geometries_from_polygon(city_boundary, tags=parking_tags)

        # Filter and keep only necessary columns
        if not parking_gdf.empty:
            parking_gdf = parking_gdf[['geometry'] + [col for col in parking_gdf.columns if col.startswith('amenity') or col.startswith('building')]]

            # Calculate total parking area
            total_parking_area = calculate_parking_area(parking_gdf)
            print(f"Total parking area for {city}: {total_parking_area} square meters")

        # Store the results
        parking_results[city] = parking_gdf

    except Exception as e:
        print(f"Could not retrieve parking data for {city}: {e}")
        parking_results[city] = None

# Save data to file
for city, gdf in parking_results.items():
    if gdf is not None and not gdf.empty:
        gdf.to_file(f"{city.replace(', ', '_').replace(' ', '_')}_parking.geojson", driver="GeoJSON")

# Combine all data into one DataFrame if needed
all_parking_gdf = pd.concat([gdf for gdf in parking_results.values() if gdf is not None and not gdf.empty])
# all_parking_gdf.to_file("all_cities_parking.geojson", driver="GeoJSON")

print("Parking data for all cities retrieved and saved successfully.")
```
